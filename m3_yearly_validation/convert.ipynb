```python
import pandas as pd
import os

def series_to_csv_files(df, output_dir=".", date_col="start_timestamp", 
                       series_col="series_name", value_col="series_value"):
    """
    Convert each series in dataframe to separate CSV files
    
    Parameters:
    -----------
    df : pandas.DataFrame
        DataFrame containing multiple time series
    output_dir : str
        Directory to save CSV files
    date_col : str
        Name of the date/timestamp column
    series_col : str
        Name of the series identifier column
    value_col : str
        Name of the values column
    """
    
    # Create output directory if it doesn't exist
    os.makedirs(output_dir, exist_ok=True)
    
    # Get unique series
    unique_series = df[series_col].unique()
    
    print(f"Converting {len(unique_series)} series to CSV files...")
    
    for series_name in unique_series:
        # Filter data for this series
        series_data = df[df[series_col] == series_name].copy()
        
        # Sort by date to ensure proper order
        series_data = series_data.sort_values(date_col)
        
        # Create new dataframe with desired columns
        csv_data = pd.DataFrame({
            'date': series_data[date_col],
            'OT': series_data[value_col]
        })
        
        # Reset index
        csv_data = csv_data.reset_index(drop=True)
        
        # Create filename (replace any problematic characters)
        filename = f"{series_name}.csv"
        filename = filename.replace("/", "_").replace("\\", "_")  # Handle any path separators
        
        # Save to CSV
        filepath = os.path.join(output_dir, filename)
        csv_data.to_csv(filepath, index=False)
        
        print(f"Saved {series_name}: {len(csv_data)} observations -> {filepath}")
    
    print(f"\nAll series saved to: {output_dir}/")
    return output_dir

# Example usage with your m3_month dataframe
def convert_m3_series():
    """
    Convert M3 monthly series to individual CSV files in current directory
    """
    # Save directly in current directory
    series_to_csv_files(
        df=m3_month,
        output_dir=".",  # Current directory
        date_col="start_timestamp",
        series_col="series_name",
        value_col="series_value"
    )
    
    print("All CSV files saved in current directory")

# Alternative: If you want to customize the date format
def series_to_csv_with_date_format(df, output_dir=".", date_format="%Y-%m-%d"):
    """
    Same as above but with custom date formatting
    """
    os.makedirs(output_dir, exist_ok=True)
    unique_series = df['series_name'].unique()
    
    print(f"Converting {len(unique_series)} series to CSV files with date format {date_format}...")
    
    for series_name in unique_series:
        series_data = df[df['series_name'] == series_name].copy()
        series_data = series_data.sort_values('start_timestamp')
        
        # Format dates if needed
        if pd.api.types.is_datetime64_any_dtype(series_data['start_timestamp']):
            formatted_dates = series_data['start_timestamp'].dt.strftime(date_format)
        else:
            formatted_dates = series_data['start_timestamp']
        
        csv_data = pd.DataFrame({
            'date': formatted_dates,
            'OT': series_data['series_value']
        })
        
        filename = f"{series_name}.csv"
        filepath = os.path.join(output_dir, filename)
        csv_data.to_csv(filepath, index=False)
        
        print(f"Saved {series_name}: {len(csv_data)} observations -> {filepath}")
    
    return output_dir

# Quick function to check one CSV file
def check_csv_file(series_name, output_dir="."):
    """
    Check what one of the generated CSV files looks like
    """
    filepath = os.path.join(output_dir, f"{series_name}.csv")
    if os.path.exists(filepath):
        df = pd.read_csv(filepath)
        print(f"Contents of {series_name}.csv:")
        print(df.head())
        print(f"Shape: {df.shape}")
        return df
    else:
        print(f"File {filepath} not found")
        return None
```

```python
# Save CSV files in current directory
series_to_csv_files(m3_month)
```
